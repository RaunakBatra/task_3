```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df=pd.read_csv('recipe_final (1).csv')
```

```python
vectorizer=TfidfVectorizer()
X_ingredients= vectorizer.fit_transform(df['ingredients_list'])
```

```python
scaler= StandardScaler()
x_numerical= scaler.fit_transform(df[['fat','calories','protein','cholesterol','sodium','fiber','carbohydrates']])
```

```python
combined= np.hstack([X_ingredients.toarray(),x_numerical])
```

```python
knn= NearestNeighbors(metric='euclidean',n_neighbors=3)
knn.fit(combined)
```

```python
def recommendation_recipes(input_features):
  input_features_scaled= scaler.transform([input_features[:7]])
  input_ingredients_transformed=vectorizer.transform([input_features[7]])
  input_combined=np.hstack([input_features_scaled,input_ingredients_transformed.toarray()])
  distances, indices=knn.kneighbors(input_combined)
  recommended_recipes=df.iloc[indices[0]]
  return recommended_recipes[['recipe_name','ingredients_list','image_url']]
```

```python
input_features=[ 15, 36, 1, 42, 21, 81, 2, 'pork belly, smoked paprika, kosher salt']
```

```python
recommendation_recipes(input_features)
```
